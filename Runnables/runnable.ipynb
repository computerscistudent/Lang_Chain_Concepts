```python
import random
class llm:
    def __init__(self):
        print("llm initialized")

    def predict(self, prompt):
        response_list = [
            "Delhi is the capital of India!!",
            "IPL is a cricket league",
            "AI stands for artificial intelligence"
        ]

        res =  random.choice(response_list)
        return {'response': res}
```

```python
llm = llm()
```

```python
result = llm.predict("capital of india?")
print(result)
```

```python
class NakliPromptTemplate :
    def __init__(self, template : str, input_variable : list):
        self.template = template
        self.input_variable = input_variable
        print("Templat class created")

    def format(self, input_dict:dict):
        return self.template.format(**input_dict)    
```

```python
template = NakliPromptTemplate( template="write a {length} poem about {topic}", input_variable=["topic"])
```

```python
prompt = template.format({'topic': 'india', 'length':'short'})
print(prompt)
```

```python
class NaklillmChain :
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt
        print("Chain Initialized!!")

    def run(self, input_dict):
        final_prompt = self.prompt.format(input_dict)  
        res = self.llm.predict(final_prompt)
        return res['response']
```

```python
chin = NaklillmChain(llm, template)
result = chin.run({'topic': 'india', 'length':'short'})
print(result)
```
